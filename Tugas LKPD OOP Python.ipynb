```python
# ==========================================
# LATIHAN 1: MEMBUAT CLASS HERO
# ==========================================

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name          # Nama Hero
        self.hp = hp              # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    def info(self):
        print(f"│ 👤 Hero: {self.name:<10} │ HP: {self.hp:<5} │ Power: {self.attack_power:<5} │")

# -- Main Program --
print("┌──────────────────────────────────────────────────┐")
print("│            LATIHAN 1: KONDISI AWAL               │")
print("├──────────────────────────────────────────────────┤")
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

hero1.info()
hero2.info()
print("└──────────────────────────────────────────────────┘")

# --- Uji Coba Tugas Analisis 1 ---
print("\n┌──────────────────────────────────────────────────┐")
print("│        LATIHAN 1: SETELAH HP DIBUAT 500          │")
print("├──────────────────────────────────────────────────┤")
hero1.hp = 500
hero1.info()
print("└──────────────────────────────────────────────────┘")

"""
======================================================================
HASIL ANALISIS LATIHAN 1
======================================================================
1. Pengamatan Output:
   Atribut `hp` berhasil diubah secara bebas dari nilai 100 menjadi 500
   melalui perintah `hero1.hp = 500`.

2. Masalah Kunci:
   Atribut pada class Hero masih bersifat PUBLIC. Siapa saja (termasuk
   pihak luar atau hacker) dapat memanipulasi status internal objek
   tanpa adanya batasan/validasi.

3. Kesimpulan:
   Sifat public pada data sensitif seperti HP berisiko tinggi menyebabkan
   kecurangan (cheat). Oleh karena itu, kita memerlukan mekanisme
   Enkapsulasi (Encapsulation) untuk mengamankan data.
======================================================================
"""
```

```python
# ==========================================
# LATIHAN 2: INTERAKSI ANTAR OBJEK (METHOD)
# ==========================================

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"│ 👤 {self.name:<10} │ HP: {self.hp:<5} │ Power: {self.attack_power:<5} │")

    def serang(self, lawan):
        print(f"⚔️  {self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"💥 {self.name} terkena damage {damage}. Sisa HP: {self.hp}\n")

# -- Main Program --
hero1 = Hero("Layla", 100, 15)
hero2 = Hero("Zilong", 120, 20)

print("┌──────────────────────────────────────────────────┐")
print("│         LATIHAN 2: PERTARUNGAN DIMULAI           │")
print("└──────────────────────────────────────────────────┘")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas

"""
======================================================================
HASIL ANALISIS LATIHAN 2
======================================================================
1. Pertanyaan Tugas:
   Mengapa parameter `lawan` pada method `serang` harus menerima
   sebuah objek utuh, bukan sekadar string nama?

2. Analisis Logika:
   Jika parameter `lawan` hanya berupa string (misal: "Zilong"), kita
   hanya bisa mencetak namanya. Kita TIDAK BISA mengurangi sisa HP
   Zilong atau memanggil method `diserang()` milik Zilong.

3. Kesimpulan:
   Dengan mengirimkan Objek Utuh sebagai parameter, method `serang`
   mendapatkan akses penuh terhadap seluruh properti dan perilaku
   objek target (membaca `lawan.name`, membaca `lawan.attack_power`,
   serta mengeksekusi `lawan.diserang()`).
======================================================================
"""
```

```python
# ==========================================
# LATIHAN 3: PEWARISAN (INHERITANCE)
# ==========================================

class Hero:
    def __init__(self, name, hp, attack_power):
        self.name = name
        self.hp = hp
        self.attack_power = attack_power

    def info(self):
        print(f"│ 🛡️  {self.name:<10} [Hero] │ HP: {self.hp:<5} │ Power: {self.attack_power:<5} │")

    def serang(self, lawan):
        print(f"⚔️  {self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"💥 {self.name} terkena damage {damage}. Sisa HP: {self.hp}\n")

class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"│ 🔮 {self.name:<10} [Mage] │ HP: {self.hp:<5} │ Mana : {self.mana:<5} │")

    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"🔥 {self.name} menyemburkan FIREBALL ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)
        else:
            print(f"❌ {self.name} gagal skill! Mana tidak cukup.\n")

# -- Main Program --
print("┌──────────────────────────────────────────────────┐")
print("│            LATIHAN 3: CLASS MAGE (CHILD)         │")
print("├──────────────────────────────────────────────────┤")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
print("└──────────────────────────────────────────────────┘\n")

print("--- AKSI PERTANDINGAN ---")
eudora.serang(balmond)         # Serangan biasa
eudora.skill_fireball(balmond) # Skill khusus Mage

"""
======================================================================
HASIL ANALISIS LATIHAN 3
======================================================================
1. Percobaan Menghapus `super().__init__()`:
   Jika baris `super().__init__()` dihapus/dijadikan komentar, program
   akan Error: `AttributeError: 'Mage' object has no attribute 'name'`.

2. Mengapa Error Terjadi?
   Class anak (`Mage`) meng-override (menimpa) constructor milik induk
   (`Hero`). Jika `super()` tidak dipanggil, proses pembuatan atribut
   `name`, `hp`, dan `attack_power` yang didefinisikan di class `Hero`
   tidak akan pernah dieksekusi.

3. Kesimpulan Peran `super()`:
   Fungsi `super()` bertindak sebagai jembatan yang memanggil logika
   inisialisasi milik Parent Class, sehingga Child Class dapat mewarisi
   seluruh atribut/method dasar tanpa perlu menulis ulang kodenya.
======================================================================
"""
```

```python
# ==========================================
# LATIHAN 4: ENKAPSULASI (MENGAMANKAN DATA HP)
# ==========================================

class Hero:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        self.__hp = hp_awal # Private Attribute

    def get_hp(self):
        return self.__hp

    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0
        elif nilai_baru > 1000:
            print("⚠️ [PROTEKSI] Cheat Terdeteksi! HP dimaksimalkan ke 1000.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"💥 {self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)

print("┌──────────────────────────────────────────────────┐")
print("│          LATIHAN 4: UJI VALIDASI SETTER          │")
print("└──────────────────────────────────────────────────┘")
print("👉 Mengubah HP ke -50 secara terstruktur...")
hero1.set_hp(-50)
print(f"✅ Sisa HP Akhir (Dinetralkan): {hero1.get_hp()}\n")

print("👉 Uji coba hacking via Name Mangling...")
print(f"⚠️ Nilai yang dibaca paksa: {hero1._Hero__hp}")

"""
======================================================================
HASIL ANALISIS LATIHAN 4
======================================================================
1. Percobaan Akses Paksa (Name Mangling):
   - Nilai HP tetap BISA MUNCUL saat dipanggil dengan `hero1._Hero__hp`.
   - Penyebab: Python mengubah nama atribut private secara internal
     menjadi `_NamaClass__namaAtribut` (Name Mangling).
   - Alasan larangan: Meskipun bisa diakses, tindakan ini melanggar
     konvensi OOP dan merusak proteksi internal class yang dapat
     memicu bug pada program.

2. Uji Hilangnya Logika Validasi Setter:
   - Jika logika `if` dan `elif` dihapus, perintah `set_hp(-100)` akan
     membuat HP hero menjadi -100 (nilai yang tidak valid dalam game).

3. Kesimpulan:
   Method Setter bertindak sebagai "gerbang filter" yang memastikan
   data tetap valid dan aman sebelum disimpan ke dalam objek.
======================================================================
"""
```

```python
# ==========================================
# LATIHAN 5: ABSTRACTION & INTERFACE (MEMBUAT KONTRAK)
# ==========================================

from abc import ABC, abstractmethod

class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"🗡️  Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"│ 🦸 Type: Hero    │ Name: {self.nama:<15} │ Status: Ready  │")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"🐾 Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"│ 👾 Type: Monster │ Kind: {self.jenis:<15} │ Status: Hostile│")

# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")

print("┌──────────────────────────────────────────────────┐")
print("│         LATIHAN 5: IMPLEMENTASI INTERFACE        │")
print("├──────────────────────────────────────────────────┤")
h.info()
m.info()
print("└──────────────────────────────────────────────────┘")

"""
======================================================================
HASIL ANALISIS LATIHAN 5
======================================================================
1. Uji Melanggar Kontrak (Menghapus method serang di Class Hero):
   - Error yang muncul: `TypeError: Can't instantiate abstract class
     Hero with abstract method serang`.
   - Artinya: Python menolak membuat objek dari class `Hero` karena
     class tersebut belum mengimplementasikan semua method wajib yang
     dijanjikan pada `GameUnit`.

2. Uji Membuat Objek Langsung (`unit = GameUnit()`):
   - Error yang muncul: `TypeError: Can't instantiate abstract class GameUnit...`.
   - Mengapa Dilarang?: Abstract Class dirancang murni sebagai "blueprint"
     atau cetakan konsep, bukan objek nyata.
   - Kegunaannya: Untuk memastikan seluruh class turunan (seperti Hero
     dan Monster) memiliki method wajib yang seragam.
======================================================================
"""
```

```python
# ==========================================
# LATIHAN 6: POLYMORPHISM (FLEKSIBILITAS INTERAKSI)
# ==========================================

class Hero:
    def __init__(self, nama):
        self.nama = nama
    def serang(self):
        print("👊 Hero menyerang dengan tangan kosong.")

class Mage(Hero):
    def serang(self):
        print(f"✨ {self.nama:<10} (Mage)   -> Menembakkan Bola Api! 💥")

class Archer(Hero):
    def serang(self):
        print(f"🏹 {self.nama:<10} (Archer) -> Memanah tepat di sasaran! 🎯")

class Fighter(Hero):
    def serang(self):
        print(f"⚔️  {self.nama:<10} (Fighter)-> Memukul dengan pedang! 🗡️")

class Healer(Hero):
    def serang(self):
        print(f"❇️  {self.nama:<10} (Healer) -> Menyembuhkan area sekitar! 💚")

# -- Penerapan Polymorphism --
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord"),
    Healer("Rafaela")
]

print("┌──────────────────────────────────────────────────┐")
print("│            LATIHAN 6: PERANG DIMULAI             │")
print("└──────────────────────────────────────────────────┘")
for pahlawan in pasukan:
    pahlawan.serang()

"""
======================================================================
HASIL ANALISIS LATIHAN 6
======================================================================
1. Uji Skalabilitas (Menambah Class Healer):
   - Program berjalan lancar tanpa perlu mengubah kode perulangan `for`.
   - Keuntungan: Programmer dapat menambahkan puluhan jenis hero baru
     di masa depan tanpa merusak atau mengubah kode sistem utama.

2. Uji Konsistensi Penamaan (Mengubah serang -> tembak_panah pada Archer):
   - Jika nama method diubah, perulangan `pahlawan.serang()` akan
     menjalankan method `serang()` milik class Induk (`Hero`), bukan
     method `tembak_panah()`.
   - Kesimpulan: Konsep Polimorfisme mewajibkan nama method antara Parent
     dan Child harus persis sama agar perintah tunggal dapat
     menghasilkan respon yang sesuai.
======================================================================
"""
```

```python
# ==========================================
# PROYEK INTEGRASI: MYEDOTEL (OUTPUT TABEL HTML)
# ==========================================

from abc import ABC, abstractmethod
from IPython.display import display, HTML

class KamarHotel(ABC):
    def __init__(self, nama_kamar, stok_awal, harga_dasar):
        self.nama_kamar = nama_kamar
        self.__stok = 0
        self.__harga_dasar = harga_dasar
        self.tambah_stok(stok_awal)

    def get_stok(self):
        return self.__stok

    def get_harga_dasar(self):
        return self.__harga_dasar

    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print(f"❌ [GAGAL] Stok {self.nama_kamar} tidak boleh negatif ({jumlah}).")
        else:
            self.__stok += jumlah
            print(f"✅ [BERHASIL] Stok {self.nama_kamar} bertambah +{jumlah} (Total: {self.__stok}).")

    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, stok_awal, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok_awal, harga_dasar)
        self.fasilitas = fasilitas

    def hitung_harga_total(self, jumlah_malam):
        pajak = 0.10 * self.get_harga_dasar()
        total_per_malam = self.get_harga_dasar() + pajak
        return total_per_malam * jumlah_malam, pajak

    def tampilkan_detail(self):
        return f"<b>{self.nama_kamar}</b><br><small style='color: gray;'>Fasilitas: {self.fasilitas}</small>"


class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, stok_awal, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok_awal, harga_dasar)
        self.kapasitas = kapasitas

    def hitung_harga_total(self, jumlah_malam):
        pajak = 0.05 * self.get_harga_dasar()
        total_per_malam = self.get_harga_dasar() + pajak
        return total_per_malam * jumlah_malam, pajak

    def tampilkan_detail(self):
        return f"<b>{self.nama_kamar}</b><br><small style='color: gray;'>Kapasitas: {self.kapasitas}</small>"


def proses_transaksi(daftar_pesanan):
    total_tagihan = 0

    rows_html = ""
    for idx, pesanan in enumerate(daftar_pesanan, start=1):
        kamar = pesanan["kamar"]
        malam = pesanan["malam"]

        subtotal, pajak = kamar.hitung_harga_total(malam)
        total_tagihan += subtotal

        harga_fmt = f"Rp {kamar.get_harga_dasar():,.0f}".replace(",", ".")
        pajak_fmt = f"Rp {pajak:,.0f}".replace(",", ".")
        subtotal_fmt = f"Rp {subtotal:,.0f}".replace(",", ".")

        pajak_persen = "10%" if isinstance(kamar, KamarDeluxe) else "5%"
        badge_type = "<span style='background-color:#4CAF50; color:white; padding:3px 8px; border-radius:4px; font-size:11px;'>DELUXE</span>" if isinstance(kamar, KamarDeluxe) else "<span style='background-color:#2196F3; color:white; padding:3px 8px; border-radius:4px; font-size:11px;'>STANDARD</span>"

        rows_html += f"""
        <tr>
            <td style="text-align:center;">{idx}</td>
            <td>{badge_type}<br>{kamar.tampilkan_detail()}</td>
            <td>{harga_fmt}</td>
            <td>{pajak_fmt} ({pajak_persen})</td>
            <td style="text-align:center;">{malam} Malam</td>
            <td style="font-weight:bold; color:#2E7D32;">{subtotal_fmt}</td>
        </tr>
        """

    total_tagihan_fmt = f"Rp {total_tagihan:,.0f}".replace(",", ".")

    html_code = f"""
    <div style="font-family: Arial, sans-serif; max-width: 750px; border: 1px solid #ddd; padding: 20px; border-radius: 8px; box-shadow: 0 4px 8px rgba(0,0,0,0.1);">
        <h2 style="color: #1A237E; margin-top:0; border-bottom: 2px solid #1A237E; padding-bottom: 8px;">🏨 STRUK PEMESANAN KAMAR MYEDOTEL</h2>
        <table style="width: 100%; border-collapse: collapse; margin-top: 15px;">
            <thead>
                <tr style="background-color: #f2f2f2; border-bottom: 2px solid #ccc; text-align: left;">
                    <th style="padding: 8px; text-align:center;">No</th>
                    <th style="padding: 8px;">Detail Kamar</th>
                    <th style="padding: 8px;">Harga / Malam</th>
                    <th style="padding: 8px;">Pajak</th>
                    <th style="padding: 8px; text-align:center;">Durasi</th>
                    <th style="padding: 8px;">Subtotal</th>
                </tr>
            </thead>
            <tbody>
                {rows_html}
            </tbody>
        </table>
        <hr style="border: 0; border-top: 1px dashed #ccc; margin: 20px 0;">
        <div style="text-align: right; font-size: 18px; font-weight: bold; color: #1A237E;">
            TOTAL TAGIHAN: <span style="color: #D32F2F;">{total_tagihan_fmt}</span>
        </div>
    </div>
    """
    display(HTML(html_code))


# ==========================================
# MAIN PROGRAM (SIMULASI MYEDOTEL)
# ==========================================

print("┌──────────────────────────────────────────────────┐")
print("│          SETUP KAMAR & MANAJEMEN STOK            │")
print("└──────────────────────────────────────────────────┘")
deluxe1 = KamarDeluxe("Kamar Deluxe Sea View", 0, 1500000, "Private Pool")
standard1 = KamarStandard("Kamar Standard Superior", 0, 500000, "2 Orang")

deluxe1.tambah_stok(10)
standard1.tambah_stok(-5) # Ditolak
standard1.tambah_stok(20)

keranjang_pemesanan = [
    {"kamar": deluxe1, "malam": 2},
    {"kamar": standard1, "malam": 1}
]

# Tampilkan Struk HTML Visual
proses_transaksi(keranjang_pemesanan)

"""
======================================================================
HASIL ANALISIS PROYEK INTEGRASI MYEDOTEL
======================================================================
1. Enkapsulasi:
   Atribut `__stok` dan `__harga_dasar` terlindungi. Pengisian stok
   sebesar -5 otomatis ditolak oleh validasi setter.

2. Abstraksi:
   Class `KamarHotel` menjadi standar acuan yang menjamin setiap tipe
   kamar memiliki method `tampilkan_detail()` dan `hitung_harga_total()`.

3. Pewarisan & Polimorfisme:
   `KamarDeluxe` dan `KamarStandard` mewarisi properti dasar dari
   `KamarHotel`, namun masing-masing memiliki perhitungan pajak yang
   berbeda (10% vs 5%). Fungsi `proses_transaksi()` memproses gabungan
   tipe kamar secara fleksibel tanpa memerlukan pengecekan kondisional.
======================================================================
"""
```
